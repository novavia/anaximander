```python
from typing import Optional
import anaximander as nx

@nx.compile("sqlalchemy")
class Account(nx.Model):
    """A customer account, which also maps to an application tenant."""
    name: str = nx.field(key=True)

    facilities: list["Facility"] = nx.relationship()
    groups: list["MachineGroup"] = nx.relationship()
```

```python
from abc import ABC, ABCMeta

class X(ABCMeta):
    
    def __instancecheck__(self, subclass):
        supercheck = super().__instancecheck__(subclass)
        if supercheck is True:
            return True
        else:
            return issubclass(subclass, model)


class Y(metaclass=X):

    @classmethod
    def __subclasshook__(cls, subclass):
        return True


class C(Y):
    pass

# C.register(D)
```

```python
issubclass(D, C)
```

```python
isinstance(D, X)
```
